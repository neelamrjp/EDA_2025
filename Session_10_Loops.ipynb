```python
#Case-1: when ever the correct pwd : code should stop
# Case-2: whenever you are enter the wrong
          # wrong pwd
          # still have 2 chances
          # still have 1 cahnces

# Case-3: whenerver all the chances
#         # try afer 24hours print
correct_password='python'
for i in range(3):
    password=eval(input("enter the number:"))
    if correct_password==password:
        print("correct password")
        break
    else:
         chances = 2 - i
        if chances > 0:
            print("Wrong password!")
            print("You have", chances, "chance(s) left.")
        else:
            print("Wrong password!")
            print("Try again after 24 hours.")
        
        
    
```

```python
correct_password = "python123"

for attempt in range(3):
    password = input("Enter password: ")

    if password == correct_password:
        print("Correct password! Access granted.")
        break
    else:
        chances = 2 - attempt

        if chances > 0:
            print("Wrong password!")
            print("You have", chances, "chance(s) left.")
        else:
            print("Wrong password!")
            print("Try again after 24 hours.")
```
