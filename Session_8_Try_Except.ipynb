# TYPES OF ERRORS 

- SYNTAX ERROR

- IDENTATION ERROR

- TYPE ERROR

- NAME ERROR

- ZERO DIVISION ERROR

- MODULE NOT FOUND EROOR

- ATTITRIBUTE ERROR

- VALUE ERROR

```python
- there are two blocks
    - try block (write code in try block)
    - except block (if any error occurs in try block code it will be catch in except block)
```

```python
a=10
b=20
c=a+b
print(f"the addition of {a} and {b} is {c}")
```

```python
try:
    a=10
    print(a)
    b=20
    print(b)
    c=a/0
    print(f"the addition of {a} and {b} is {c}")
except Exception as e:
    print(e)
    print("hello")
```

```python
try:
    'a'+100
except Exception as e:
    print(e)
```

```python
try:
    import random
    random.randint1()
except Exception as ee:
    print(ee)
```

```python
try:
    import random
    a=random.randint(1,100)
    print(a)
except Exception as ee:
    print(ee)
```

```python
import math
try:
    a=math.sqrt(64)
    print(a)
except Exception as ee:
    print(ee)
```

```python

try:
    import math
    a=math.sqrt(64)
    print(a)
except Exception as ee:
    print(ee)
```
