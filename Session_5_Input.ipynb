# PROBLEMS

```python
a=eval(input("enter the number1:"))
b=eval(input("enter the number2:"))
print(f"the addition of {a} and {b} is {a+b}")
print(f"the subtraction of {a} and {b} is {a-b}")
print(f"the multiplication of {a} and {b} is {a*b}")
print(f"the division of {a} and {b} is {a/b}")
```

```python
pi=eval(input("enter pi value:"))
r=eval(input("enter the r value:"))
print(f"the area of circle is {pi*r*r}")
```

```python
h=eval(input("enter the height:"))
b=eval(input("enter the breadth:"))
print(f"the area of triangle is {0.5*b*h}")
```

```python
a=eval(input("enter the number1:"))
b=eval(input("enter the number2:"))
c=eval(input("enter the number3:"))
print(f"the average of {a} ,{b} and {c} is {(a+b+c)/3}") 
```

```python
dollar=eval(input("enter the dollar value:"))
rupees=eval(input("enter the rupees value:"))
print(f"the value of {dollar}$ is {dollar*rupees} rupees")
```

```python
weight=eval(input("enter the weight:"))
pound=eval(input("enter the pound:"))
print(f"the {weight}kg is {weight*pound} in pounds ")
```

```python
bill_amount=eval(input("enter the bill amount:"))
tip_percentage=eval(input("enter the tip percentage:"))
tip_amount=bill_amount*tip_percentage/100
total_amount=bill_amount+tip_amount
print(f"the total amount is {total_amount}")
```

```python
salary=eval(input("enter the salary:"))
tax_percentage=eval(input("enter the tax percentage:"))
tax_amount=salary*tax_percentage/100
print(f"the total amount is {salary+tax_amount}")
```

```python
a,b,c=input("enter the values:").split()
print(a,b,c)
```
