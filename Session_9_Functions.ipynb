## RETURN

- function want to give something to us 

- we should save using return

```python
def average55(a,b,c):
    add=(a+b+c)/3
    avg=round(add,2)
    print(avg)
average55(10,20,30)
```

```python
def average55(a,b,c):
    add=(a+b+c)/3
    avg=round(add,2)
    print(avg)
    return(avg)
average55(10,20,30)
```

```python
def average55(a,b,c):
    add=(a+b+c)/3
    avg=round(add,2)
    print(avg)
    return(avg,add)
average55(10,20,30)
```

```python
def average55(a,b,c):
    add=(a+b)
    print(add)
    avg=(a+b+c)/3
    print(avg)
    return(avg,add)
average55(10,20,30)
```
