# TYPE CASTING

- CONVERT ONE DATA TYPE TO ANOTHER DATA TYPE

## INT TO OTHER 

```python
NUMBER=10
float(NUMBER)
```

```python
str(NUMBER)
```

```python
bool(NUMBER)
```

```python
number=-10
float(number),str(number),bool(number)
```

```python
number=0
float(number),float(number),bool(number)
```

## FLOAT TO OTHER

```python
number1=10.5
int(number1),str(number1),bool(number1)
```

```python
number1=-10.5
int(number1),str(number1),bool(number1)
```

```python
number1=0.0
int(number1),str(number1),bool(number1)
```

```python
number1=-0.0
int(number1),str(number1),bool(number1)
```

```python
number1=-1.0
int(number1),str(number1),bool(number1)
```
